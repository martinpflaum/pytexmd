```python
import os
import sys
from pathlib import Path
sys.path.append(str(Path(os.getcwd()).parent))
import subprocess
import pytexmd as ptm


input_root = "tex"
output_root = "output"
```

```python


repos = {
    "https://gitlab.com/librimathematicae/FANCyProject.git": Path(input_root) / "FANCyProject",
    "https://gitlab.com/librimathematicae/CRingProject.git": Path(input_root) / "CRingProject",
}

for repo_url, target_dir in repos.items():
    if (target_dir / ".git").exists():
        print(f"Repository already downloaded at: {target_dir.resolve()}")
    elif target_dir.exists() and any(target_dir.iterdir()):
        print(f"Directory exists and is not empty: {target_dir.resolve()}")
        print("Skipping clone to avoid overwriting existing files.")
    else:
        target_dir.parent.mkdir(parents=True, exist_ok=True)
        subprocess.run(["git", "clone", repo_url, str(target_dir)], check=True)
        print(f"Cloned into: {target_dir.resolve()}")
```

```python
from pathlib import Path

file_path = Path(input_root) / "FANCyProject" / "FANCyProject.tex"

if file_path.exists():
    print(f"✓ File exists: {file_path.resolve()}")
else:
    print(f"✗ File NOT found: {file_path.resolve()}")
```

```python
input_file_name = input_root+"/FANCyProject/FANCyProject.tex"
output_folder = output_root+"/fancyproject"
ptm.process_file(input_file_name,output_folder,depth=3)

```
